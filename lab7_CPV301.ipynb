```python
import cv2
import tkinter as tk
from tkinter import filedialog, messagebox
from tkinter import Label, Button, Frame
from PIL import Image, ImageTk

# Function to choose an image and display it
def choose_image():
    global img_path, img_display, original_image
    img_path = filedialog.askopenfilename(filetypes=[("Image Files", "*.jpg;*.jpeg;*.png")])
    if img_path:
        original_image = Image.open(img_path)
        original_image.thumbnail((500, 500))  # Limit the size of the image to 500x500 pixels
        img_display = ImageTk.PhotoImage(original_image)
        panel.config(image=img_display)
        panel.image = img_display

# Function to detect faces
def detect_faces():
    if img_path:
        # Load the chosen image
        image = cv2.imread(img_path)
        gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

        # Load the Haar Cascade
        face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')

        # Detect faces
        faces = face_cascade.detectMultiScale(gray, scaleFactor=1.1, minNeighbors=5, minSize=(30, 30))
        
        if len(faces) == 0:
            messagebox.showinfo("Result", "No faces detected in the image.")
            return
        
        # Draw rectangles around detected faces
        for (x, y, w, h) in faces:
            cv2.rectangle(image, (x, y), (x + w, y + h), (255, 0, 0), 2)
        
        # Convert image to RGB for display
        image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
        image = Image.fromarray(image)
        image.thumbnail((500, 500))  # Limit the size of the displayed image to 500x500 pixels
        img_display = ImageTk.PhotoImage(image)
        panel.config(image=img_display)
        panel.image = img_display

# Initialize the GUI
root = tk.Tk()
root.title("Face Detection Application")
root.geometry("600x700")
root.configure(bg='#f0f0f0')

# Create a frame for better layout
frame = Frame(root, bg='#ffffff', padx=10, pady=10)
frame.pack(pady=20)

# Create and place GUI elements
panel = Label(frame, bg='#e6e6e6')
panel.pack()

btn_frame = Frame(root, bg='#f0f0f0')
btn_frame.pack(pady=10)

btn_choose = Button(btn_frame, text="Choose Image", command=choose_image, bg='#4caf50', fg='white', font=('Helvetica', 12), padx=10, pady=5)
btn_choose.grid(row=0, column=0, padx=5)

btn_detect = Button(btn_frame, text="Detect Faces", command=detect_faces, bg='#2196f3', fg='white', font=('Helvetica', 12), padx=10, pady=5)
btn_detect.grid(row=0, column=1, padx=5)

# Ensure buttons stay visible after choosing an image
panel.lift()
btn_frame.lift()

# Run the GUI loop
root.mainloop()
```
